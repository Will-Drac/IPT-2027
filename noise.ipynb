# Here's how we can define noise

1. Choose a minimum and maximum frequency for your noise. The maximum is important because physical signals can't have infinite frequency and the minimum is important because the types of noise considered below can't have a zero-frequency component. We may also need to define a minimum frequency much larger than the pendulum's natural frequency.

2. Choose a frequency spacing. The noise will have one sinusoid for each frequency between min and max, spaced out by the frequency spacing. Tighter spacings mean that the noise takes longer before repeating.

2. Choose a type of noise. Generally, it takes the form $$ P(\Omega) = A \Omega^{-\alpha} $$ where $P(\Omega)$ is the power of frequency $\Omega$, $A$ is some scaling factor for the power, and $\alpha$ controls which type of noise it is. Some values of $\alpha$ have names:
   - $\alpha = -1$: Blue noise
   - $\alpha = 0$: White noise
   - $\alpha = 1$: Pink noise
   - $\alpha = 2$: Brown noise

It would be good to have something more physically meaningful than an arbitrary $A$. Let's check the total power:

$$
\begin{aligned}
P_\text{tot} &= \int_{\Omega_\text{min}}^{\Omega_\text{max}} P(\Omega) d\Omega 
\\
&= \int_{\Omega_\text{min}}^{\Omega_\text{max}} A \Omega^{-\alpha} d\Omega
\\
&\Rightarrow A = \begin{cases}
P_\text{tot} \left(\ln \frac{\Omega_\text{max}}{\Omega_\text{min}} \right)^{-1} & \alpha = 1
\\
P_\text{tot} (1-\alpha) \left(\Omega_\text{max}^{1-\alpha} - \Omega_\text{min}^{1-\alpha} \right)^{-1} & \alpha \ne 1
\end{cases}
\end{aligned}
$$

So for a desired total power of the noise, we can compute $A$ as a function of the frequency bounds and the type of noise.

4. Give each sinusoid in the frequency range a random amplitude, with mean $\sqrt{P(\Omega)}$
