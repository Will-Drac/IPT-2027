# Equations of motions

Equations of motion are similar to simple pendulum but with an added y-displacement due to the oscilating pivot. Let
$$
x(t)=l\sin(\theta)\\
y(t)=y_p(t)-l\cos(\theta) 
$$
Where $l$ is the length of the pedulum, $\theta$ the angle w.r.t. time and $y_p(t)=A\cos(\omega t)$ the displacement of the oscillating pivot with frequency $\omega$.
Plugging t=0 we find that the pivot is at $y_p(0)=A$ with $y_0$ and $x_0$ depending on $\theta (0)$ (initial conditions). 
\
The Lagrangian is given by $L=T-V$, where 
$$
T = \frac{m}{2}(\dot x^2 + \dot y^2) \\
V = mgy
$$
and the Euler-Lagrange satisfying
$$
0=\frac{d}{dt}(\frac{\partial L}{\partial \dot \theta})-\frac{\partial L}{\partial \theta}
$$
Plugging in yields the equation of motion:
$$
\ddot \theta = - \frac{\ddot y_p + g}{l}\sin \theta , \quad \ddot y_p = -A\omega^2\cos(\omega t)
$$
