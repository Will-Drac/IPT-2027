## Equation of motion derivation

For an object with mass $m$ and moment of inertia matrix $\mathrm{I}_p$ about a pivot which has velocity $\mathbf{v}_p$, where the displacement from the pivot to the center of mass is $\mathbf{r}_{\text{pivot} \rightarrow \text{com}}$, and the object is rotating with angular velocity $\omega$ about the pivot, the kinetic energy is
$$
T = \frac{1}{2} m \mathbf{v}^2_p + m \mathbf{v}_p \cdot (\mathbf{\omega} \times \mathbf{r}_{\text{pivot} \rightarrow \text{com}}) + \frac{1}{2}\mathbf{\omega} \cdot \mathrm{I}_p \mathbf{\omega} .
$$

For our pendulum, we have $\omega = \dot{\theta} \hat{k}$, so the only important component of moment of inertia is $I = \mathrm{I}_p[3,3]$. We also get $\mathbf{\omega} \times \mathbf{r}_{\text{pivot} \rightarrow \text{com}} = \dot\theta l \, \hat\theta$ where $l$ is the distance from the pivot to the center of mass. We construct the pivot so that its velocity only in the $\hat{j}$ direction, so let's call the pivot's vertical position $d$ and velocity $\dot{d}$.

We therefore have
$$
T = \frac{1}{2} m \dot{d}^2 + ml\dot{d}\dot\theta \sin\theta + \frac{1}{2}I\dot{\theta}^2 .
$$

The potential energy is
$$
V = mgy = mg(d-l\cos\theta) .
$$

Plugging these into the Euler-Lagrange equation, we get
$$
\ddot\theta + \frac{ml}{I} (g-\ddot{d})\sin\theta = 0 .
$$

## Reducing variables

If $\ddot{d} = 0$, $\omega_0^2 = \frac{mgl}{I}$ would be the natural frequency of the pendulum. Let's use that and also define $\ddot{d}_g = \ddot{d}/g$, a unitless quantity representing effective pivot acceleration relative to gravity. This lets us rewrite the equation of motion in terms of only a few variables:
$$
\ddot\theta + \omega^2_0 (1-\ddot{d}_g) \sin\theta = 0 .
$$

For the basic Kaptiza pendulum, we have the pivot simply oscillating with amplitude $A$ and frequency $\nu$: $d = A \cos(\nu t)$. Therefore
$$
\ddot{d}_g = -\frac{A}{g} \nu^2 \cos(\nu t)
$$
is the relative acceleration.
This hints towards defining an amplitude relative to gravity $A_g = A/g$, giving us the equation of motion
$$
\ddot\theta + \omega_0^2 \left(1 + A_g \nu^2 \cos(\nu t)\right)\sin\theta = 0
$$
with only three variables describing the experimental setup: $\omega_0$, $A_g$, and $\nu$. Also note that $\omega_0$ is easily measurable by keeping the pivot still and $A_g$ and $\nu$ are chosen by the experimenter.

## Stability of the Kaptiza Pendulum

Let's split $\theta$ into two parts: the fast-moving part coming directly from the vibration of the pivot $\xi$ and the rest which we'll call $\theta_0$, so that
$$
\theta = \theta_0 + \xi
$$ 
and 
$$
\ddot\theta = \ddot\theta_0 + \ddot\xi .
$$
We'll assume that the pivot's oscillation is significantly faster than the natural frequency: $\nu \gg \omega_0$. Let's also assume that the amplitude of the pivot's motion is much smaller than the distance from the pivot to the center of mass: $l \gg A$. This means that $\xi$ will be only a tiny motion. Therefore, it is a reasonable approximation to replace $\sin\theta$ with its Taylor expansion about $\theta=\theta_0$.
$$
\sin\theta \approx \sin\theta_0 + \xi\cos\theta_0
$$

The equation of motion can now be written as 
$$
\ddot\theta_0 + \ddot\xi + \omega_0^2 \left( 1 + A_g \nu^2 \cos(\nu t) \right) \left( \sin\theta_0 + \xi\cos\theta_0 \right) = 0 .
$$
We can make some more approximations now. Since $\xi$ is "fast" and $\theta_0$ is "slow", we can say $\ddot\xi \gg \ddot\theta_0$. Even though $A_g$ is assumed to be small, $\nu$ being squared means that $A_g\nu^2 \gg 1$. We also reuse the idea that $\xi$ is small, $\xi \ll 1$. Putting these together, we can remove terms and the expression simplifies to
$$
\ddot\xi + \omega_0^2 A_g \nu^2 \cos(\nu t) \sin\theta_0 = 0 .
$$
Integrating twice and assuming that $\theta_0$ is roughly constant for the time scale of interest (namely one period of the pivot's oscillation, which is very short), we get an expression for the fast motion as a function of time
$$
\xi = A_g \omega^2_0 \sin\theta_0 \cos\nu t .
$$
Now that we know $\ddot\xi$ and $\xi$, we can substitute back into the expanded equation of motion to get an equation of motion for $\theta_0$:
$$
\ddot\theta_0 - A_g \omega_0^2 \nu^2 \sin\theta_0 \cos\nu t + \omega^2_0 \left(1 + A_g \nu^2 \cos\nu t \right) \left(\sin\theta_0 + A_g \omega^2_0 \sin\theta_0 \cos\theta_0 \cos\nu t \right) = 0 ,
$$
or, expanding:
$$
\ddot\theta_0 - A_g \omega_0^2 \nu^2 \sin\theta_0 \cos\nu t + \omega_0^2\sin\theta_0 + A_g \omega_0^4 \sin\theta_0 \cos\theta_0 \cos\nu t + A_g \omega_0^2 \nu^2 \sin\theta_0 \cos\nu t + A_g^2 \omega_0^4 \nu^2 \sin\theta_0 \cos\theta_0 \cos^2\nu t = 0 .
$$

The next approximation trick is to time-average this equation over a period of the pivot's oscillation. Computationally, this means integrating from some time $t'$ to $t' + \frac{2\pi}{\nu}$ then dividing by the time difference $\frac{2\pi}{\nu}$. Since $\theta_0$ moves very slowly relative to this time scale, we can say that $\theta_0$ and $\ddot\theta_0$ are constant for this integration. There are two important identities to note:
$$
\int_{t'}^{t'+\frac{2\pi}{\nu}} \cos(\nu t) dt = 0
$$
and
$$
\int_{t'}^{t'+\frac{2\pi}{\nu}} \cos^2(\nu t) dt = \frac{1}{2}\cdot\frac{2\pi}{\nu} .
$$
So integrating the above expression, the result is
$$
\begin{aligned}
\ddot\theta_0 + \omega_0^2 \sin\theta_0 + \frac{1}{2} A_g^2 \omega_0^4 \nu^2 \sin\theta_0 \cos\theta_0 &= 0 ,
\\
\ddot\theta_0 + \omega_0^2 \left(1 + \frac{1}{2}A_g^2 \omega_0^2 \nu^2 \cos\theta_0 \right) \sin\theta_0 &= 0 .
\end{aligned}
$$
This is now an equation for the general motion of the pendulum, and notice it has no time dependence! We can check its stability in the inverted position $\theta_0 = \pi$. When $\theta_0 \approx \pi$, $\sin\theta_0 \approx -\theta_0$ and $\cos\theta_0 \approx -1$. Therefore,
$$
\ddot\theta_0 - \omega_0^2 (1 - \frac{1}{2}A_g^2 \omega_0^2 \nu^2) \theta_0 = 0
$$
which means the pendulum is stable when
$$
\begin{aligned}
- \omega_0^2 (1 - \frac{1}{2}A_g^2 \omega_0^2 \nu^2) &> 0
\\
\implies \omega_0^2 \nu^2 A_g^2 &> 2 .
\end{aligned}
$$

## Stability of Kaptiza's pendulum driven by any periodic function

As found before, the general equation of motion for a pendulum with an accelerating pivot is
$$
\ddot\theta + \omega^2_0 (1-\ddot{d}_g) \sin\theta = 0 .
$$
Previously, we said that the pivot is oscillating sinusoidally, so $\ddot{d}_g = -A_g \nu^2 \cos\nu t$. Let's instead drive the pivot by a more general function made up of sinusoids:
$$
d_g = \sum^{\infty}_{n=0} A_n\sin(n\nu_1 t) + B_n\cos(n\nu_1 t) .
$$
This is a Fourier series, and therefore it can represent any periodic function. For brevity, I'll write $\nu_n = n\nu_1$ and $\mathbf{d}_n = A_n\sin(\nu_n t) + B_n\cos(\nu_n t) $. We can write the second derivative as
$$
\ddot{d}_g = -\sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n .
$$

Substituting into the equation of motion:
$$
\ddot\theta + \omega_0^2 (1 + \sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n) \sin\theta = 0 .
$$
We can then do the exact same $\xi$ trick as before, using the same assumptions. Namely, we must have that $\sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n \ll 1$. Also, even the lowest frequency $\nu_1 \gg \omega_0$. We end up with a similar long expression for $\theta_0$:
$$
\ddot\theta_0 - \omega_0^2 \left( \sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n\right ) \sin\theta_0 + \omega_0^2 \sin\theta_0 + \omega_0^4 \left( \sum^\infty_{n=0} \mathbf{d}_n \right) \sin\theta_0\cos\theta_0 + \omega_0^2 \left(\sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n\right) \sin\theta_0 + \omega_0^4 \left(\sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n\right) \left( \sum^\infty_{m=0} \mathbf{d}_m \right) \sin\theta_0 \cos\theta_0 = 0 .
$$
We can how do the same averaging trick. To get convenient results from the integrals, we must integrate across a period which is an integer multiple of the period of each sinusoid in the sum. This is where it becomes important that $d_g$ is a Fourier *series*. There exists such a period, and it's $T = \frac{2\pi}{\nu_1}$. For sinusoid $n$ with frequency $\nu_n = n\nu_1$, the period is $T_n = \frac{2\pi}{n\nu_1}$. Multiplying by integer $n$, notice we get $nT_n = T$. We can use some nice observations for the integrals:
$$
\begin{aligned}
\int_{t'}^{t'+T} \sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n dt &= \sum^\infty_{n=0} \nu_n^2 \int_{t'}^{t'+T} \mathbf{d}_n dt
\\
&= \sum^\infty_{n=0} \nu_n^2 \left( A_n \int_{t'}^{t'+T} \sin(\nu_n t) dt + B_n \int_{t'}^{t'+T} \cos(\nu_n t) dt \right)
\\
&= 0 .
\end{aligned}
$$
And for the last term,
$$
\begin{aligned}
\int_{t'}^{t'+T} \left( \sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n \right) \left( \sum^\infty_{m=0} \mathbf{d}_m \right) dt &= \int_{t'}^{t'+T} \sum^\infty_{n=0} \sum^\infty_{m=0} \nu_n^2 \mathbf{d}_n  \mathbf{d}_m  dt
\\
&=  \sum^\infty_{n=0} \sum^\infty_{m=0} \nu_n^2 \left( \int_{t'}^{t'+T} \mathbf{d}_n \mathbf{d}_m dt \right)
\end{aligned}
$$
We can greatly simplify that integral:
$$
\begin{aligned}
\int_{t'}^{t'+T} \mathbf{d}_n \mathbf{d}_m dt &= A_n A_m \int_{t'}^{t'+T} \sin\nu_n t \sin\nu_m t dt 
\\
&+ A_n B_m \int_{t'}^{t'+T} \sin\nu_n t \cos\nu_m t dt 
\\
&+ B_n A_m \int_{t'}^{t'+T} \cos\nu_n t \sin\nu_m t dt 
\\
&+ B_n B_m \int_{t'}^{t'+T} \cos\nu_n t \cos\nu_m t dt
\\
&= A_n A_m \frac{T}{2} \delta_{n,m} + 0 + 0 + B_n B_m \frac{T}{2} \delta_{n,m}
\end{aligned}
$$
$$
\begin{aligned}
\implies \int_{t'}^{t'+T} \left( \sum^\infty_{n=0} \nu_n^2 \mathbf{d}_n \right) \left( \sum^\infty_{m=0} \mathbf{d}_m \right) dt &= \sum^\infty_{n=0} \sum^\infty_{m=0} \nu_n^2 \left( A_n A_m \frac{T}{2} \delta_{n,m} + B_n B_m \frac{T}{2} \delta_{n,m} \right)
\\
&= \frac{T}{2} \sum^\infty_{n=0} \nu_n^2 \left( A_n^2 + B_n^2 \right)
\end{aligned}
$$
We can define $R_n^2 = A_n^2 + B_n^2$ to be the amplitude of the $\mathbf{d}_n$ term. Putting everything back into the $\theta_0$ equation:
$$
\ddot\theta_0 + \omega_0^2 \left( 1 + \frac{1}{2} \omega_0^2 \cos\theta_0 \sum_{n=0}^\infty \nu_n^2 R_n^2 \right) \sin\theta_0 = 0 .
$$

Similarly to before, we get stability if
$$
\omega_0^2 \sum_{n=0}^\infty \nu_n^2 R_n^2 > 2 .
$$

### Parseval's Theorem

Parseval's theorem states that, for a series like this,
$$
\frac{1}{T} \int_0^T [x(t)]^2 dt = \frac{B_0^2}{4} + \frac{1}{2} \sum^\infty_{n=1} R_n^2 .
$$
Now let $x' = \frac{dx}{dt}$, and notice:
$$
\frac{d}{dt} \left( \sum_{n=0}^\infty A_n \sin(\nu_n t) + B_n \cos(\nu_n t) \right) = \sum_{n=1}^\infty A_n \nu_n \cos(\nu_n t) - B_n \nu_n \sin(\nu_n t) .
$$
Applying Parseval's theorem to $x'$, where $R^{'2}_n = A_n^2 \nu_n^2 + B_n^2 \nu_n^2 = \nu_n^2 R_n^2$,
$$
\begin{aligned}
\frac{1}{T} \int_0^T [x'(t)]^2 dt &= 0 + \frac{1}{2} \sum_{n=1}^\infty \nu_n^2 R_n^2
\\
\implies \sum_{n=0}^\infty \nu_n^2 R_n^2 &= \frac{2}{T} \int_0^T [x'(t)]^2 dt .
\end{aligned}
$$
Notice that $\sum_{n=1}^\infty \nu_n^2 R_n^2 = \sum_{n=0}^\infty \nu_n^2 R_n^2$ since the $n=0$ term is $0$ anyway. This is an easier way to calculate the bound for any periodic function with period $T$!

$$
\omega_0^2 \int_0^T (\dot{d}_g)^2 dt > T
$$

### Example 1: Triangle waves

We know that the frequencies at which a sine wave makes a stable pendulum is $$\nu > \frac{\sqrt{2}}{\omega_0 A_g}.$$ What about at triangle wave?

A triangle wave with peak value $A_g$ and frequency $\nu$ has the Fourier series representation
$$
d_g = \frac{8A_g}{\pi^2} \sum^\infty_{n=1,3,5,\dots} \frac{(-1)^{(n-1)/2}}{n^2} \cos(n\nu t)
$$
so we have $A_n = 0$, $B_n^2 = \left(\frac{8A_g}{\pi^2 n^2}\right)^2$ and therefore
$$
\begin{aligned}
\sum^\infty_{n=0} \nu_n^2 R_n^2 &= \sum^\infty_{n=1,3,5,\dots} n^2 \nu^2 \frac{64 A_g^2}{\pi^4 n^4}
\\
&= \frac{64 A_g^2 \nu^2}{\pi^4} \sum^\infty_{n=1,3,5,\dots} \frac{1}{n^2}
\\
&= \frac{64 A_g^2 \nu^2}{\pi^4} \cdot \frac{\pi^2}{8}
\\
&= \frac{8 A_g^2 \nu^2}{\pi^2}
\end{aligned}
$$
and so the stability condition is
$$
\begin{aligned}
\omega_0^2 \frac{8 A_g^2 \nu^2}{\pi^2} &> 2
\\
\nu &> \frac{\pi}{2 \omega_0 A_g} .
\end{aligned}
$$

Note that $\pi/2$ is slightly greater than $\sqrt{2}$. The Kapitza pendulum should be slightly more stable when the pivot is oscillating with a sine wave than with a triangle wave!
